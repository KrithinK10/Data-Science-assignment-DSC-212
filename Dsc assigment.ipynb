```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

G = nx.karate_club_graph()
print(G.number_of_edges())
print(G.number_of_nodes())
```

```python
print("The neighbours of 0 are",list(G.neighbors(0)))
print("The number of neighbours of 0 are",len(list(G.neighbors(0))))
```

```python
pos = nx.spring_layout(G,seed = 50,weight = None)
plt.figure(figsize=(15,8))
nx.draw_networkx(
    G,
    pos = pos,
    with_labels = True,
    node_color = "skyblue",
    node_size = 500,
    edge_color = "gray",
    font_size = 10
)
plt.title("Karate club graph without modifications")
plt.axis("off")
plt.show()
    
```

```python
nodes = list(G.nodes())
A = nx.to_numpy_array(G,nodelist = nodes,weight = None)
print("Shape of A is",A.shape)
print(A[:5,:5])
k = A.sum(axis=1)
print("degree",k)
m = G.number_of_edges()
print(m)
n = k.sum()
print(n)
expected = np.outer(k,k)/(2*m)
print(expected)
B = A - expected 
print(B)
plt.hist([v for i,v in nx.degree(G)]);
```
